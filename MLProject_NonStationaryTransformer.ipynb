```python
# from google.colab import drive
# drive.mount('/content/drive')
# file_path = '/content/drive/MyDrive/Machine Learning Assignments/Project/ETTm2.csv'

# Find current path
import os
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from sklearn.cluster import DBSCAN
from sklearn.preprocessing import StandardScaler

os = os.getcwd()
print(os)

file_path = './ETDataset/ETT-small/ETTm2.csv'

df = pd.read_csv(file_path)

df.head()
```

## Normalizing

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
df_scaled = scaler.fit_transform(df.iloc[:, 1:])
df_scaled = pd.DataFrame(df_scaled, columns=df.columns[1:])

df_scaled.head()
```

```python
# Plot a sample of date and output variables

x = df['date'].apply(lambda x: pd.to_datetime(x))

output_variables = ['OT', 'HUFL', 'HULL', 'MUFL', 'MULL', 'LUFL', 'LULL']

for variable in output_variables:
  y = df_scaled[variable]

  plt.figure(figsize=(12, 4))

  plt.plot(x, y)
  plt.grid()
  plt.xlabel('Date')
  plt.ylabel(variable)

  #Print just some ticks
  plt.xticks(x[::5000], rotation=30)

  plt.title(f'Date vs {variable}')
  plt.show()

```

## Printing the difference between two consecutives values

```python
from sklearn.preprocessing import StandardScaler

for variable in output_variables:
  y = df_scaled[variable]
  y_diff = np.diff(y)
  y_diff_abs = np.abs(y_diff)

  scaler = StandardScaler()
  y_diff_abs = scaler.fit_transform(y_diff_abs.reshape(-1, 1))
  if variable == 'HUFL':
    Y_HUFL_diff = y_diff_abs
  plt.figure(figsize=(12, 4))

  plt.plot(x[1:], y_diff_abs)
  plt.grid()
  plt.xlabel('Date')
  plt.ylabel(variable)

  #Print just some ticks
  plt.xticks(x[::5000], rotation=30)

  plt.title(f'Date vs {variable} Diff')
  plt.show()
```

## Classifying the abrupt changes per variable with sliding window


```python
eps_values = [0.1, 0.3, 0.5, 0.8, 1.0]
min_samples_values = [1, 2, 3, 5, 8]
# variable = 'HUFL'
for variable in output_variables:

  # 2) sliding window feature extraction
  seq_len = 96
  win_len = seq_len
  step = seq_len

  y = df_scaled[variable]
  count = 0
  for eps in eps_values:
    for min_samples in min_samples_values:
      outlier_samples_all = []
      for start in range(0, len(y) - win_len + 1, step):
          w = y[start:start+win_len]

          w_diff = np.diff(w)
          w_diff_abs = np.abs(w_diff)
          w_diff_abs_max = np.max(w_diff_abs)

          scaler = StandardScaler()
          X = scaler.fit_transform(w_diff_abs.reshape(-1, 1))

          cl = DBSCAN(eps=eps, min_samples=min_samples)
          labels = cl.fit_predict(X)

          outlier_windows_idx = np.where(labels == -1)[0]

          for idx in outlier_windows_idx:
            global_index = start + idx

            outlier_samples_all.append([global_index, start])


      outlier_samples = [x[0] for x in outlier_samples_all]

      outlier_samples = np.unique(outlier_samples)
      print(len(outlier_samples))

      if len(outlier_samples) > 0:
        print("EPS: ", eps, " MIN_SAMPLES: ", min_samples)
        plt.figure(figsize=(12, 4))

        plt.plot(x, y)
        plt.scatter(x[outlier_samples], y[outlier_samples], color='red', s=20, label='DBSCAN outlier (change)')
        plt.grid()
        plt.xlabel('Date')
        plt.ylabel(variable)
        plt.title(f'{variable} DBSCAN-based abrupt change detection (DBSCAN)')
        plt.legend()
        plt.tight_layout()
        #Print just some ticks
        plt.xticks(x[::5000], rotation=30)
        plt.show()
      else:
        print("No abrupt changes found")
```

```python
dbscan_params = {'OT':   [0.5, 3],
                 'HUFL': [0.8, 3],
                 'HULL': [0.3, 2],
                 'MUFL': [0.3, 2],
                 'MULL': [0.8, 3],
                 'LUFL': [0.8, 3],
                 'LULL': [1.0, 3]

                 }

# Initialize empty lists for each row
df['is_abrupt'] = [[] for _ in range(len(df))]
df['dbscan_start'] = [[] for _ in range(len(df))]

for params in dbscan_params:
  eps = dbscan_params[params][0]
  min_samples = dbscan_params[params][1]

  print("Testing eps: ", eps, " min_samples: ", min_samples, "for variable: ", params)

  y = df_scaled[params]
  y_diff = np.diff(y)
  y_diff_abs = np.abs(y_diff)

  outlier_samples_all = []
  for start in range(0, len(y) - win_len + 1, step):
      w = y[start:start+win_len]

      w_diff = np.diff(w)
      w_diff_abs = np.abs(w_diff)
      w_diff_abs_max = np.max(w_diff_abs)

      scaler = StandardScaler()
      X = scaler.fit_transform(w_diff_abs.reshape(-1, 1))

      cl = DBSCAN(eps=eps, min_samples=min_samples)
      labels = cl.fit_predict(X)

      outlier_windows_idx = np.where(labels == -1)[0]

      for idx in outlier_windows_idx:
        global_index = start + idx

        # Append variable name and start index
        df.at[global_index, 'is_abrupt'].append(params)
        df.at[global_index, 'dbscan_start'].append(start)

        outlier_samples_all.append([global_index, start])


  # Convert outlier_samples_all to a dict for fast lookup
  # # {global_index: start}
  # abrupt_dict = {idx: start for idx, start in outlier_samples_all}

  # # Create the columns
  # df['is_abrupt'] = df.index.map(lambda i: 1 if i in abrupt_dict else 0)
  # df['dbscan_start'] = df.index.map(lambda i: abrupt_dict.get(i, np.nan))


  outlier_samples = [x[0] for x in outlier_samples_all]

  outlier_samples = np.unique(outlier_samples)
  print("Number of abrupt changes: " ,len(outlier_samples))

  if len(outlier_samples) > 0:
    print("EPS: ", eps, " MIN_SAMPLES: ", min_samples)
    plt.figure(figsize=(12, 4))

    plt.plot(x, y)
    plt.scatter(x[outlier_samples], y[outlier_samples], color='red', s=20, label='DBSCAN outlier (change)')
    plt.grid()
    plt.xlabel('Date')
    plt.ylabel(params)
    plt.title(f'{params} DBSCAN-based abrupt change detection (DBSCAN)')
    plt.legend()
    plt.tight_layout()
    #Print just some ticks
    plt.xticks(x[::5000], rotation=30)
    plt.show()
  else:
    print("No abrupt changes found")

```
